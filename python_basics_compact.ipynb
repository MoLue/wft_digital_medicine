## 1. Python Variables and Data Types

In Python, you store information in **variables**. Think of them as labeled boxes that hold different types of data. Let's look at examples from a medical context:

```python
# Numeric data types
patient_age = 45                  # Integer (whole number)
body_temperature = 37.2           # Float (decimal number)

# String data type (text)
patient_name = "Jane Doe"
diagnosis = "Type 2 Diabetes"

# Boolean data type (True/False)
is_admitted = True
has_allergies = False

# Print the variables and their types
print(f"Patient: {patient_name}, Type: {type(patient_name)}")
print(f"Age: {patient_age}, Type: {type(patient_age)}")
print(f"Temperature: {body_temperature}°C, Type: {type(body_temperature)}")
print(f"Diagnosis: {diagnosis}")
print(f"Admitted: {is_admitted}")
```

## 2. Data Structures: Lists and Dictionaries

### Lists
Lists are ordered collections. In medicine, you might use them to store a series of measurements.

```python
# Blood glucose readings (mg/dL) over a week
glucose_readings = [95, 105, 110, 98, 102, 115, 107]

# Access elements (indexing starts at 0!)
print(f"First reading: {glucose_readings[0]} mg/dL")
print(f"Last reading: {glucose_readings[-1]} mg/dL")

# Basic statistics
average = sum(glucose_readings) / len(glucose_readings)
print(f"Average: {average:.1f} mg/dL")
print(f"Min: {min(glucose_readings)}, Max: {max(glucose_readings)} mg/dL")

# Add a new reading
glucose_readings.append(101)
print(f"Updated list: {glucose_readings}")
```

### Dictionaries
Dictionaries store key-value pairs — perfect for structured data like patient records.

```python
# A patient record as a dictionary
patient = {
    "id": "P12345",
    "name": "John Smith",
    "age": 58,
    "diagnosis": "Hypertension",
    "medications": ["Lisinopril", "Hydrochlorothiazide"],
    "vital_signs": {
        "blood_pressure": "140/90",
        "heart_rate": 72
    }
}

# Access values
print(f"Patient: {patient['name']}")
print(f"Diagnosis: {patient['diagnosis']}")
print(f"Medications: {', '.join(patient['medications'])}")
print(f"Heart Rate: {patient['vital_signs']['heart_rate']} bpm")

# Add new information
patient["allergies"] = ["Penicillin"]
print(f"Allergies: {patient['allergies']}")
```

## 3. Functions

Functions let you package reusable code. Here's a practical medical example:

```python
def calculate_bmi(weight_kg, height_m):
    """Calculate Body Mass Index."""
    bmi = weight_kg / (height_m ** 2)
    return bmi


def interpret_bmi(bmi):
    """Interpret BMI according to WHO classification."""
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal weight"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"


# Test the functions
weight = 70  # kg
height = 1.75  # m

bmi = calculate_bmi(weight, height)
category = interpret_bmi(bmi)

print(f"Weight: {weight} kg, Height: {height} m")
print(f"BMI: {bmi:.1f}")
print(f"Category: {category}")
```

## 4. Working with Data using Pandas

Pandas is the go-to library for data analysis in Python. It lets you work with tabular data (like Excel spreadsheets) efficiently.

Cheat Sheet: [Pandas Cheat Sheet (PDF)](https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf)

```python
import pandas as pd

# Load a synthetic diabetes dataset from GitHub
url = "https://raw.githubusercontent.com/MoLue/wft_digital_medicine/main/data/synthetic_diabetes.csv"
df = pd.read_csv(url)

# First look at the data
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

```python
# Get an overview of the dataset
df.info()
```

```python
# Summary statistics
df.describe()
```

```python
# Filtering: Find patients with high blood glucose (> 150 mg/dL)
high_glucose = df[df["BloodGlucose"] > 150]
print(f"Patients with high blood glucose: {len(high_glucose)} out of {len(df)}")
high_glucose.head()
```

```python
# Grouping: Average blood glucose by diabetes status
print("Average values by Diabetes Status:")
df.groupby("DiabetesStatus")[["BloodGlucose", "BodyMassIndex", "PatientAge"]].mean()
```

## 5. Data Visualization

Visualizations help you understand data patterns at a glance. We use **matplotlib** and **seaborn** for plotting.

Cheat Sheet: [Seaborn Charts](https://www.kaggle.com/code/themlphdstudent/cheat-sheet-seaborn-charts)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a nice style
sns.set(style="whitegrid")

# Histogram: Distribution of Blood Glucose levels
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="BloodGlucose", hue="DiabetesStatus", bins=20, palette="coolwarm")
plt.title("Distribution of Blood Glucose by Diabetes Status")
plt.xlabel("Blood Glucose (mg/dL)")
plt.ylabel("Count")
plt.show()
```

```python
# Scatter plot: BMI vs Blood Glucose
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="BodyMassIndex",
    y="BloodGlucose",
    hue="DiabetesStatus",
    palette="coolwarm",
    alpha=0.7
)
plt.title("BMI vs. Blood Glucose")
plt.xlabel("Body Mass Index")
plt.ylabel("Blood Glucose (mg/dL)")
plt.show()
```
